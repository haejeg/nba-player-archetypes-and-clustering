```python
print(df.columns) # Hate these dataframe column names
df = df.rename(columns={
    '2P': 'fga%_2P', '0-3': 'fga%_0-3', '3-10': 'fga%_3-10', '10-16': 'fga%_10-16', '16-3P': 'fga%_16-3P', '3P': 'fga%_3P',
    '2P.1': 'fg%_2P', '0-3.1': 'fg%_0-3', '3-10.1': 'fg%_3-10', '10-16.1': 'fg%_10-16', '16-3P.1': 'fg%_16-3P', '3P.1': 'fg%_3P',
    '2P.2': 'ast%_2P', '3P.2': 'ast%_3P',
    '%FGA': 'dunk_fga%', '#': 'dunks_made',
    '%3PA': 'corner_3pa%', '3P%': 'corner_3p%',
    'Att.': 'heave_att', 'Md.': 'heave_made',
    '-9999': 'player_id',
})
print(df.columns)
df
```

df

```python
totals = pd.read_csv('../../data/basketball-reference/players(25-26)/regular-season/totals.csv')
totals = totals.rename(columns={'Player-additional': 'player_id'})

# Join on player_id AND Team so traded players' per-team rows line up correctly
df = df.merge(totals[['player_id', 'Team', 'FGA']], on=['player_id', 'Team'], how='left')

zones = ['0-3', '3-10', '10-16', '16-3P', '3P']
for z in zones:
    df[f'fga_per36_{z}'] = df[f'fga%_{z}'] * df['FGA'] / df['MP'] * 36
```

```python
# Rim attempts (0-3 ft) per 36 minutes, players with 500+ MP
rim = df[df['MP'] >= 500][['Player', 'Team', 'MP', 'FGA', 'fga%_0-3', 'fga_per36_0-3']]
rim.sort_values('fga_per36_0-3', ascending=False).head(20)
```
